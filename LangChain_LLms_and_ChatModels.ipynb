### Accessing Open source LLMs with HuggingFace


- Here we accessing Microsoft Phi-3-mini-4k instruct .It is state of the art open model.

- We are also going to see google gemma 2b model also

```python
from langchain_community.llms import HuggingFaceEndpoint

PHI3_URL="https://api-inference.huggingface.co/models/microsoft/Phi-3-mini-4k-instruct"

phi3_params={"wait_for_model":True,
             "do_sample":False,
             "return_full_text":False,
             "max_new_tokens":1000,
}


llm = HuggingFaceEndpoint(endpoint_url=PHI3_URL,
                          task="text_generation",
                          **phi3_params)

```

```python
prompt="Explain what is  Artificial Neural Network"
```

```python
## In phi3 input prompt is formatted in a specific way

phi3_prompt="""<|user|>Explain what is artificial neural Network in five bullet points<|end|>
<|assistant|>"""
print(phi3_prompt)
```

```python
response=llm.invoke(phi3_prompt)
print(response)
```

```python
#accessin google gemma 2B instruct

gemma_url="https://api-inference.huggingface.co/models/google/gemma-1.1-2b-it"

gemma_params={"wait_for_model":True,
             "do_sample":False,
             "return_full_text":False,
             "max_new_tokens":1000,
}

llm=HuggingFaceEndpoint(
    endpoint_url=gemma_url,
    task="text_generation",
    **gemma_params
)
```

```python
prompt
```

```python
response = llm.invoke(prompt)
print(response)
```

### Conversational Prompting With ChatGPT  

```python
from langchain_openai import ChatOpenAI

llm=ChatOpenAI(model_name="gpt-3.5-turbo",temperature=0)
```

```python
from langchain_core.messages import HumanMessage,SystemMessage

prompt="Explain Generative AI in 3 bullet points"
system_prompt="Act as helpful assistant and give a single example along with the description "

messages=[SystemMessage(content=system_prompt),
          HumanMessage(content=prompt),]


messages

```

```python
response=llm.invoke(messages)
response
```

```python
print(response.content)
```

```python
### adding the past conversation history into messages

messages.append(response)

### adding new prompt to conversational history list
prompt="Explain the last point in detail "

messages.append(HumanMessage(content=prompt))

messages
```

```python
response=llm.invoke(messages)
print(response.content)
```

### Conversational Promptinng with Open Source LLms with HuggingFace

```python
from langchain_community.llms import HuggingFaceEndpoint

gemma_api_url="https://api-inference.huggingface.co/models/google/gemma-1.1-2b-it"


gemma_params={ "wait_for_model":True,
              "do_sample":False,
               "return_full_text":False,
               "max_new_tokens":1000,
}


llm=HuggingFaceEndpoint(endpoint_url=gemma_api_url,
                        task="conversation",
                        **gemma_params

)

```

```python
llm
```

```python
from langchain_community.chat_models import ChatHuggingFace

chat_gemma=ChatHuggingFace(llm=llm,model_id='google/gemma-1.1-2b-it')
```

```python
prompt="Explain generative Ai in five points along with examples "


messages=[(HumanMessage(content=prompt)),]

response=chat_gemma.invoke(messages)

print(response.content)
```

```python
## adding response conversation in message history

messages.append(response)

messages
```

```python
new_prompt="Explain 4th point in detail "
messages.append(HumanMessage(content=new_prompt))

response=chat_gemma.invoke(messages)
response
```

```python
print(response.content)
```
